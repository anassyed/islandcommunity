# 🏝️ The Island Community Skills Analysis

This notebook analyzes survey responses from "The Island: Building Our Community" to understand the skills, interests, and potential roles of community members.

## 📋 Survey Overview

The survey uses a scenario-based assessment where members choose their **MOST** and **LEAST** appealing tasks across 12 different scenarios. Responses are mapped to 12 skill categories:

- **(A) Operational/Scribe** - Documentation, planning, scheduling
- **(B) Outdoor** - Exploration, hiking, nature activities
- **(C) Kids' Activities** - Games, supervision, child-focused events
- **(D) Physical/Sports** - Fitness, sports, physical activities
- **(E) Educational** - Teaching, learning, skill development
- **(F) Social** - Gatherings, potlucks, mingling
- **(G) Community Service** - Caring for others, cleanup, serving
- **(H) Spiritual** - Prayer, halaqa, spiritual support
- **(I) Leadership/Strategy** - Decision-making, governance, planning
- **(J) Artisan/Engineer** - Building, fixing, crafting
- **(K) Arts & Culture** - Entertainment, creative expression
- **(L) Economy/Resource** - Resource management, rationing, barter

## 🎯 Scoring System

- **MOST appealing choice**: +2 points
- **LEAST appealing choice**: -1 point

This creates a range from -10 to +20 for each category (10 questions × ±2 or ±1 points).

```python
# Import required libraries
import pandas as pd
import numpy as np
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 50)

print("✅ Libraries imported successfully!")
```

## 📥 Load Survey Data

```python
# Load the CSV file
df = pd.read_csv('The Island_ Our New Community (Responses) - Form Responses 1.csv')

# Clean column names (remove extra whitespace)
df.columns = df.columns.str.strip()

print(f"📊 Loaded {len(df)} responses")
print(f"📋 Columns: {len(df.columns)}")
print("\n🔍 First few responses:")
df.head(3)
```

## 🗺️ Question-to-Category Mapping

Each survey option maps to one of the 12 skill categories. We'll create a comprehensive mapping based on the question structure.

```python
# Define the 12 categories
categories = {
    'A': 'Operational',
    'B': 'Outdoor',
    'C': 'Kids Activities',
    'D': 'Physical/Sports',
    'E': 'Educational',
    'F': 'Social',
    'G': 'Community Service',
    'H': 'Spiritual',
    'I': 'Leadership/Strategy',
    'J': 'Artisan/Engineer',
    'K': 'Arts & Culture',
    'L': 'Economy/Resource'
}

# Question mapping - maps each choice to its category
# Based on the survey structure from The_island_community.md
question_mapping = {
    # Question 1: The First 24 Hours
    'Finalizing the schedule': 'A',  # Operational
    'Leading an exploration team': 'B',  # Outdoor
    'Planning and leading simple games': 'C',  # Kids Activities
    'Organizing a daily "fitness" team': 'D',  # Physical/Sports
    
    # Question 2: First Week Priorities
    'Attending a "how-to" session': 'E',  # Educational
    'Hosting a low-key "story time"': 'F',  # Social
    'Coordinating a 3-hour shift': 'G',  # Community Service
    'Finding a clean, quiet space and organizing the daily communal prayers': 'H',  # Spiritual
    
    # Question 3: Establishing Order
    'Leading the main "council" meeting': 'I',  # Leadership
    'Taking charge of the salvaged tools': 'J',  # Artisan/Engineer
    'Writing a "song of the island"': 'K',  # Arts & Culture
    'Designing a "barter" system': 'L',  # Economy/Resource
    
    # Question 4: Building Morale
    'Managing the "check-in" desk': 'A',  # Operational
    'Inviting a guest speaker': 'E',  # Educational
    'Do a research to teach': 'E',  # Educational (variant)
    'Developing the "rules" for the camp': 'I',  # Leadership
    'Planning a community picnic': 'F',  # Social
    
    # Question 5: Governance & Crafting
    'Organizing a community-wide potluck': 'F',  # Social
    'Figuring out how to build a better shelter': 'J',  # Artisan/Engineer
    'Supervising a "play zone"': 'C',  # Kids Activities
    'Organizing a "community cleanup"': 'G',  # Community Service
    
    # Question 6: Long-Term Projects
    'Organizing a "talent show"': 'K',  # Arts & Culture
    'Joining a friendly, low-stakes sports': 'D',  # Physical/Sports
    'Leading a short, reflective discussion': 'H',  # Spiritual
    'Managing the central store': 'L',  # Economy/Resource
    
    # Question 7: Skills & Society
    'Writing down the "official" camp rules': 'A',  # Operational
    'Setting up a casual "board game night"': 'F',  # Social
    'Carving decorations': 'K',  # Arts & Culture
    'Helping to set up and run a "story time"': 'C',  # Kids Activities
    
    # Question 8: Community Care
    'Taking a "how-to" class': 'E',  # Educational
    'Spending the day carving new tools': 'J',  # Artisan/Engineer
    'Setting up a small, fun "sports day"': 'D',  # Physical/Sports
    'Deciding the "value" of rare salvaged items': 'L',  # Economy/Resource
    
    # Question 9: Daily Life
    'Making the final decision on *where* to build': 'I',  # Leadership
    'Running a donation drive': 'G',  # Community Service
    'Leading a guided "discovery" walk': 'B',  # Outdoor
    'Taking the time to provide one-on-one spiritual': 'H',  # Spiritual
    
    # Question 10: The Future
    'Organizing the master spreadsheet': 'A',  # Operational
    'Keeping the "master account"': 'L',  # Economy/Resource
    'Volunteering to lead a group in a shared moment of Dhikr': 'H',  # Spiritual
    'Watching an informative documentary': 'E',  # Educational
    
    # Question 11: Community Life (additional mappings)
    'Organizing a group outing': 'F',  # Social
    'Helping to serve meals': 'G',  # Community Service
    'Starting the "official" written journal': 'K',  # Arts & Culture
    'Leading a morning group': 'D',  # Physical/Sports
    
    # Question 12: The Future (additional mappings)
    'Organizing a simple, fun activity for children': 'C',  # Kids Activities
    'Thinking about the "1-year plan"': 'I',  # Leadership
    'Designing and leading the construction': 'J',  # Artisan/Engineer
    'Simply making sure to welcome new faces': 'F',  # Social
    
    # Additional variations
    'Creating the step-by-step plan': 'A',  # Operational
}

print("✅ Category mapping created")
print(f"📊 {len(categories)} categories defined")
print(f"🗺️ {len(question_mapping)} question options mapped")
```

## 🧮 Score Calculation Function

```python
def categorize_response(response_text):
    """Map a response text to its category using fuzzy matching."""
    if pd.isna(response_text):
        return None
    
    response_text = str(response_text)
    
    # Try exact matching first
    for key, category in question_mapping.items():
        if key.lower() in response_text.lower():
            return category
    
    # Fallback: check for category labels in parentheses
    for cat_code, cat_name in categories.items():
        if f"({cat_name})" in response_text or f"({cat_code})" in response_text:
            return cat_code
    
    return None


def calculate_individual_scores(row):
    """Calculate scores for an individual based on their MOST/LEAST choices."""
    scores = {cat: 0 for cat in categories.keys()}
    
    # Process each question (20 columns: 10 questions × 2 columns each)
    for col in row.index:
        if 'MOST Appealing' in col:
            category = categorize_response(row[col])
            if category:
                scores[category] += 2  # +2 for MOST appealing
        elif 'LEAST Appealing' in col:
            category = categorize_response(row[col])
            if category:
                scores[category] -= 1  # -1 for LEAST appealing
    
    return pd.Series(scores)


print("✅ Scoring functions defined")
```

## 🎯 Process All Responses

```python
# Clean up the name column
df['Name'] = df['Name'].str.strip()

# Calculate scores for each person
score_df = df.apply(calculate_individual_scores, axis=1)

# Add names
score_df.insert(0, 'Name', df['Name'].values)

# Replace category codes with full names for readability
score_df.columns = ['Name'] + [categories[col] for col in score_df.columns[1:]]

print("✅ Scores calculated for all members")
print(f"\n📊 Score Summary:")
score_df
```

## 👤 Individual Profiles

Let's identify each member's top strengths and least preferred areas.

```python
def get_individual_profile(name, top_n=3):
    """Get an individual's top strengths and weaknesses."""
    person_data = score_df[score_df['Name'] == name].iloc[0]
    scores = person_data.drop('Name').sort_values(ascending=False)
    
    profile = {
        'name': name,
        'top_strengths': scores.head(top_n).to_dict(),
        'least_preferred': scores.tail(top_n).to_dict()
    }
    
    return profile


# Generate profiles for all members
profiles = []
for name in score_df['Name'].dropna():
    if name:  # Skip empty names
        profile = get_individual_profile(name)
        profiles.append(profile)
        
        print(f"\n🎯 **{profile['name']}**")
        print("   Top Strengths:")
        for skill, score in profile['top_strengths'].items():
            print(f"      • {skill}: {score:+d}")
        print("   Least Preferred:")
        for skill, score in profile['least_preferred'].items():
            print(f"      • {skill}: {score:+d}")

print(f"\n✅ Generated {len(profiles)} individual profiles")
```

## 📊 Community-Wide Analysis

```python
# Calculate aggregate statistics
category_columns = [col for col in score_df.columns if col != 'Name']
community_stats = score_df[category_columns].agg(['mean', 'median', 'std', 'min', 'max'])

print("📈 Community-Wide Statistics:")
print(community_stats.round(2))

# Identify high-interest and low-interest categories
avg_scores = score_df[category_columns].mean().sort_values(ascending=False)

print("\n🔥 Most Popular Categories (by average score):")
for cat, score in avg_scores.head(5).items():
    print(f"   {cat}: {score:.2f}")

print("\n❄️ Least Popular Categories:")
for cat, score in avg_scores.tail(5).items():
    print(f"   {cat}: {score:.2f}")
```

## 📊 Visualization 1: Community Heat Map

## 📊 Visualization 2: Category Distribution

## 🕸️ Visualization 3: Individual Radar Charts

View detailed profiles for each member.

```python
def create_radar_chart(name):
    """Create a radar chart for an individual member."""
    person_data = score_df[score_df['Name'] == name].iloc[0]
    scores = person_data.drop('Name')
    
    fig = go.Figure()
    
    fig.add_trace(go.Scatterpolar(
        r=scores.values,
        theta=scores.index,
        fill='toself',
        name=name,
        line=dict(color='#1f77b4', width=2),
        fillcolor='rgba(31, 119, 180, 0.3)'
    ))
    
    fig.update_layout(
        polar=dict(
            radialaxis=dict(
                visible=True,
                range=[-10, 20]
            )
        ),
        title=f'🎯 Skills Profile: {name}',
        showlegend=False,
        height=500
    )
    
    return fig
```

```python


# Create radar charts for first 3 members as examples
sample_names = score_df['Name'].dropna()
for name in sample_names:
    if name:
        fig = create_radar_chart(name)
        fig.show()

print(f"\n✅ Sample radar charts created")
print(f"💡 Use create_radar_chart('Name') to view any member's profile")
```

## 👥 Team Formation Suggestions

Identify potential teams based on complementary skills.

```python
def suggest_teams_by_category(category, team_size=3):
    """Suggest top performers for a specific category."""
    top_performers = score_df.nlargest(team_size, category)[['Name', category]]
    return top_performers


print("🎯 Suggested Teams by Category:\n")

# Create teams for key categories
key_categories = ['Spiritual', 'Leadership/Strategy', 'Educational', 
                  'Community Service', 'Social', 'Artisan/Engineer']

for cat in key_categories:
    print(f"\n**{cat} Team:**")
    team = suggest_teams_by_category(cat, team_size=3)
    for idx, row in team.iterrows():
        print(f"   • {row['Name']} (Score: {row[cat]:+d})")

print("\n✅ Team suggestions generated")
```

## 📉 Gap Analysis

Identify skill categories that need more volunteers.

```python
# Count how many people have positive scores in each category
positive_counts = (score_df[category_columns] > 0).sum().sort_values(ascending=True)

# Count how many people have strong interest (score >= 4)
strong_interest_counts = (score_df[category_columns] >= 4).sum().sort_values(ascending=True)

gap_analysis = pd.DataFrame({
    'Category': positive_counts.index,
    'Interested Members': positive_counts.values,
    'Highly Interested': strong_interest_counts.values,
    'Avg Score': [score_df[cat].mean() for cat in positive_counts.index]
})

print("🔍 Gap Analysis - Categories Needing Attention:\n")
print(gap_analysis.to_string(index=False))

print("\n⚠️ Categories with <5 interested members:")
low_interest = gap_analysis[gap_analysis['Interested Members'] < 5]
for _, row in low_interest.iterrows():
    print(f"   • {row['Category']}: Only {row['Interested Members']} members")

print("\n✅ Gap analysis complete")
```

## 📤 Export Results

```python
# Export individual scores
score_df.to_csv('community_scores.csv', index=False)
print("✅ Exported: community_scores.csv")

# Export team recommendations
team_recommendations = []
for cat in category_columns:
    top_3 = suggest_teams_by_category(cat, 3)
    for idx, row in top_3.iterrows():
        team_recommendations.append({
            'Category': cat,
            'Member': row['Name'],
            'Score': row[cat]
        })

teams_df = pd.DataFrame(team_recommendations)
teams_df.to_csv('team_recommendations.csv', index=False)
print("✅ Exported: team_recommendations.csv")

# Export gap analysis
gap_analysis.to_csv('gap_analysis.csv', index=False)
print("✅ Exported: gap_analysis.csv")

print("\n🎉 All exports complete!")
```
